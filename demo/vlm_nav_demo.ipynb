# ROAM: VLM terrain maps on your own photos

[![Open In Colab](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/github/leggedrobotics/roam/blob/main/demo/vlm_nav_demo.ipynb)

Upload a photo. A vision-language model labels the terrain (risk, steppable, preferred area), and SAN
open-vocabulary segmentation turns the labels into per-pixel maps, as in
[ROAM](https://roam-hiking.github.io/) ([paper](https://doi.org/10.1126/scirobotics.aee1549),
[arXiv](https://arxiv.org/abs/2611.25215)).

1. **Runtime → Change runtime type → T4 GPU**.
2. Optional: add a Colab secret `GEMINI_API_KEY` ([free key](https://aistudio.google.com/apikey)).
   You can also paste the key in the app. The examples work without one.
3. **Runtime → Run all**. The last cell prints a `gradio.live` link: open it.

The first run takes about 5 minutes (it builds detectron2).
